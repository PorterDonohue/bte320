```python
Number_1 = float(input("Number 1: "))
sign = input("Sign")
Number_2 = float(input("Number 2: "))

if sign == "+":
  print(Number_1 + Number_2)
if sign == "-":
  print(Number_1 - Number_2)
if sign == "*":
  print(Number_1 * Number_2)
if sign == "/":
  print(Number_1 / Number_2)
```

```python
num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number: "))
op = input("Choose numerical operation (e.g., +, -, *, etc.): ")

res = None # Initialize res to None

if op == "+":
  res = num1 + num2
elif op == "-":
  res = num1 - num2
elif op == "*":
  res = num1 * num2
elif op == "/":
  res = num1 / num2
elif op == "//":
  res = num1 // num2
elif op == "**":
  res = num1 ** num2
elif op == "%":
  res = num1 % num2
else:
  print("Invalid operation")

if res is not None: # Only print if res was assigned a value
  print(f"Result: {res}")
```

```python
print("Hour", "Fee")
for hour in range(1,9):
  fee = 5 + 2.5 * hour
  if fee < 10.0:
    fee = 10.0
  elif fee > 20.0:
    fee = 20.0
  print(f"{hour}    {fee}")
```
