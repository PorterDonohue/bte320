```python
Number_1 = float(input("Number 1: "))
sign = input("Sign")
Number_2 = float(input("Number 2: "))

if sign == "+":
  print(Number_1 + Number_2)
if sign == "-":
  print(Number_1 - Number_2)
if sign == "*":
  print(Number_1 * Number_2)
if sign == "/":
  print(Number_1 / Number_2)
```

```python
num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number: "))
op = input("Choose numerical operation (e.g., +, -, *, etc.): ")

res = None # Initialize res to None

if op == "+":
  res = num1 + num2
elif op == "-":
  res = num1 - num2
elif op == "*":
  res = num1 * num2
elif op == "/":
  res = num1 / num2
elif op == "//":
  res = num1 // num2
elif op == "**":
  res = num1 ** num2
elif op == "%":
  res = num1 % num2
else:
  print("Invalid operation")

if res is not None: # Only print if res was assigned a value
  print(f"Result: {res}")
```

```python
print("Hour", "Fee")
for hour in range(1,9):
  fee = 5 + 2.5 * hour
  if fee < 10.0:
    fee = 10.0
  elif fee > 20.0:
    fee = 20.0
  print(f"{hour}    {fee}")
```

```python
print("Hour", "Fee")
hour = 1
while hour < 9: # or while hour <= 8
  fee = 5 + 2.5 * hour
  if fee < 10.0:
    fee = 10.0
  elif fee > 20.0:
    fee = 20.0
  print(f"{hour}    {fee}")
  hour += 1
```

```python
import math

ticket_price = 10
initial_attendees = 20

print("Ad$\tProfit")
for ad in range(0, 201, 25):
  revenue = (initial_attendees + 2 * round(math.sqrt(ad))) * ticket_price
  costs = 200 + ad
  profit = revenue - costs
  print(f"{float(ad)}\t{float(profit)}")
```

```python
import math

ticket_price = 10
initial_attendees = 20

print("Ad$\tProfit")
ad = 0
while ad < 201:
  revenue = (initial_attendees + 2 * round(math.sqrt(ad))) * ticket_price
  costs = 200 + ad
  profit = revenue - costs
  print(f"{float(ad)}\t{float(profit)}")
  ad += 25
```

Assignment 5

```python
import random

check = input("Do you want to play a game (y/n)? ")

while check == 'y':
  player = input("Choose b/w rock, paper, scissors: ")
  computer = random.choice(['rock', 'paper', 'scissors'])

  if player == computer:
    print('Draw')
  elif player == 'rock' and computer == 'scissors':
    print('Player wins')
  elif player == 'scissors' and computer == 'paper':
    print('Player wins')
  elif player == 'paper' and computer == 'rock':
    print('Player wins')
  else:
    print('Computer wins')

  check = input("Do you want to play again (y/n)? ")
```

```python
numbers = []

num = int(input("Enter a number: "))
while num != 0:
  numbers.append(0)
  num = int(input("Enter another number: "))
numbers.sort(reverse=True)
print(numbers)
```

```python
numbers = []

while True:
  num = int(input("Enter another number: "))
  numbers.append(num)
  if num == 0:
    break

numbers = numbers[:-1]
numbers.sort(reverse=True)
print(numbers)
```
