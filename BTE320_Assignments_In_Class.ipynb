```python
def loan_payment(principal, annual_rate, months):
  assert (isinstance(principal, (int, float)) and
          isinstance(annual_rate, (int, float)) and
          isinstance(months, int)), "All inputs must be numbers, months an integer"

  if (principal <= 0) or (annual_rate < 0) or (months <= 0):
    raise ValueError("Invalid Input: Principal and months must be positive, annual_rate non-negative.")

  try:
    r = annual_rate / 12
    if r == 0:
      pmt = principal / months
    else:
      pmt = r * principal / (1 - (1 + r)**(-months))
    return pmt
  except Exception as e:
    print(f"Error calculating loan payment: {e}")
```

```python
def loan_payment(principle: float, annual_rate: float, months: int) -> float:
  assert isinstance(principle, (int, float)), "Principle must be numeric"
  assert isinstance(annual_rate, (int, float)), "Annual rate must be numeric"
  assert isinstance(months, int), "Months must be an integer"

  if principle <= 0:
    raise ValueError("Principle must be > 0")
  if annual_rate < 0:
    raise ValueError("Annual rate cannot be negative")
  if months <= 0:
    raise ValueError("Number of months must be > 0")

  r = annual_rate / 12
  n = months
  P = float(principle)

  if r == 0:
    return P/n
  else:
    return r*P / (1-(1+r)**(-n))

if __name__ == "__main__":
  try:
    P = float(input("Enter principle: "))
    apr = float(input("Enter annual interest rate as a decimal: "))
    n = int(input("Enter number of months: "))
    payment = loan_payment(P, apr, n)
    print(f"Monthly payment: {payment: .2f}")
  except Exception as e:
    print(e)
```

```python
def getValue(n):
  assert isinstance(n, int), "Input variable 'n' must be an interger" #or: assert type(n) == int
  assert n > 0
  items = []

  for i in range(n):
    items.append(float(input("Value: ")))

  return items

def calcAverage(alist):
  assert isinstance(alist, list)
  assert len(alist) > 2
  try:
    return sum(alist[2:])/len(alist[2:])
  except Exception as e:
    print(f"Error: {e}")
    raise ValueError("The length of list 'alist' must be greater than 2")

def main():
  n = int(input("Enter the number of list items: "))
  alist = getValue(n)
  avg = calcAverage(alist)
  print(avg)

main()
```

```python
class Student:
  def __init__(self, name, number):
    self.__name = name
    self.__number = number
    self.__courses = []

  def get_name(self):
    return self.__name

  def get_number(self):
    return self.__number

  def get_courses(self):
    return self.__courses

  def set_name(self, newName):
    self.__name = newName

  def __len__(self):
    return len(self.__courses)

  def enroll(self, newCourse):
    if newCourse not in self.__courses:
      self.__courses.append(newCourse)
      print(f"Enrolled in {newCourse}")
    else:
      print(f"Student {self.__name} is already enrolled in {newCourse}")
```

```python
s = Student("Jane Doe", 12345)
print(f"Student name: {s.get_name()}")
print(f"Show currently enrolled courses: {s.get_courses()}")
s.enroll('BTE 320')
print(f"Show currently enrolled courses: {s.get_courses()}")
s.enroll('BTE 320')
print(f"Total number of courses enrolled: {len(s)}")
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    dx = (self.x - other.x)**2
    dy = (self.y - other.y)**2
    return (dx + dy)**0.5

class Rectangle:
  def __init__(self, p1, p2, p3, p4):

    self.p1 = Point(0, 0)
    self.p2 = Point(5, 0)
    self.p3 = Point(0, 5)
    self.p4 = Point(5, 5)
    self.width = self.p1.distance(self.p2)
    self.height = self.p1.distance(self.p3)

  def width_(self):
    return self.p1.distance(self.p2)

  def height_(self):
    return self.p1.distance(self.p3)

  def area(self):
    return self.width * self.height


p1 = Point(0,0)
p2 = Point(5,0)
p3 = Point(0,5)
p4 = Point(5,5)
z = Rectangle(p1, p2, p3, p4)
print(z.area())
```
